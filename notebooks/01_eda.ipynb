```python
# training and testing

X = df.drop(columns=["Churn", "customerID"])
y = df["Churn"]
```

```python
print(X.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
numerical_features = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]
```

```python
# converting categorical to numerical by one -hot encoding
# keeping numerical as it is
categorical_features = [
    col for col in X_train.columns
    if col not in numerical_features
]

print("Numerical:", numerical_features)
print("Categorical:", categorical_features)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numerical_features
        ),
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ]
)
```

```python
preprocessor.fit(X_train)
```

```python
#logistic regression
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression())
    ]
)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print(y_pred[:10])
```

```python
#model evaluation
from sklearn.metrics import (accuracy_score,confusion_matrix,classification_report)

print("Accuracy:", accuracy_score(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
y_prob = model.predict_proba(X_test)[:, 1]
```

```python
y_pred_04 = [
    "Yes" if prob >= 0.4 else "No"
    for prob in y_prob
]
```

```python
print("Threshold = 0.4")
print(classification_report(y_test, y_pred_04))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_04))
```

```python
from sklearn.metrics import roc_auc_score

roc_auc = roc_auc_score(y_test, y_prob)

print("ROC-AUC:", roc_auc)
```

```python
from sklearn.metrics import roc_curve
import matplotlib.pyplot as plt

fpr, tpr, thresholds = roc_curve(y_test, y_prob, pos_label="Yes")

plt.plot(fpr, tpr, label=f"Logistic Regression (AUC = {roc_auc:.3f})")
plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate (Recall)")
plt.title("ROC Curve")
plt.legend()
plt.show()
```

```python
from sklearn.metrics import precision_recall_curve, average_precision_score

precision, recall, thresholds = precision_recall_curve(
    y_test,
    y_prob,
    pos_label="Yes"
)

avg_precision = average_precision_score(
    y_test,
    y_prob,
    pos_label="Yes"
)

print("Average Precision:", avg_precision)
```

```python
plt.plot(
    recall,
    precision,
    label=f"Logistic Regression (AP = {avg_precision:.3f})"
)

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.legend()
plt.show()
```

```python
# decision tree model
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

tree_preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ],
    remainder="passthrough"
)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.pipeline import Pipeline

tree_model = Pipeline(
    steps=[
        ("preprocessor", tree_preprocessor),
        ("classifier", DecisionTreeClassifier(random_state=42))
    ]
)
```

```python
tree_model.fit(X_train, y_train)
```

```python
tree_pred = tree_model.predict(X_test)
```

```python
print("Training Accuracy:", tree_model.score(X_train, y_train))
print("Testing Accuracy:", tree_model.score(X_test, y_test))
```

```python
tree_model.fit(X_train, y_train)

print("Training Accuracy:", tree_model.score(X_train, y_train))
print("Testing Accuracy:", tree_model.score(X_test, y_test))
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "classifier__max_depth": [3, 5, 7, 10],
    "classifier__min_samples_split": [2, 10, 20],
    "classifier__min_samples_leaf": [1, 5, 10]
}
```

```python
grid_search = GridSearchCV(
    tree_model,
    param_grid,
    cv=5,
    scoring="f1_macro",
    n_jobs=-1
)
```

```python
grid_search.fit(X_train, y_train)
```

```python
print("Best Parameters:")
print(grid_search.best_params_)

print("Best CV Score:")
print(grid_search.best_score_)
```

```python
best_tree = grid_search.best_estimator_

best_tree_pred = best_tree.predict(X_test)

print(classification_report(y_test, best_tree_pred))
```

```python
#random forest
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

rf_model = Pipeline(
    steps=[
        ("preprocessor", tree_preprocessor),
        (
            "classifier",
            RandomForestClassifier(
                n_estimators=100,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)
```

```python
rf_model.fit(X_train, y_train)
```

```python
rf_pred = rf_model.predict(X_test)

print("Training Accuracy:", rf_model.score(X_train, y_train))
print("Testing Accuracy:", rf_model.score(X_test, y_test))
```

```python
print("Best parameters:")
print(rf_grid.best_params_)

print("Best CV macro F1:")
print(rf_grid.best_score_)
```

```python
best_rf = rf_grid.best_estimator_

rf_pred = best_rf.predict(X_test)

print("Training Accuracy:", best_rf.score(X_train, y_train))
print("Testing Accuracy:", best_rf.score(X_test, y_test))

print(classification_report(y_test, rf_pred))
```

```python
rf_prob = best_rf.predict_proba(X_test)[:, 1]

from sklearn.metrics import roc_auc_score, average_precision_score

print(
    "RF ROC-AUC:",
    roc_auc_score(y_test, rf_prob)
)

print(
    "RF Average Precision:",
    average_precision_score(
        y_test,
        rf_prob,
        pos_label="Yes"
    )
)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

for threshold in [0.30, 0.35, 0.40, 0.45, 0.50, 0.55, 0.60]:

    predictions = [
        "Yes" if prob >= threshold else "No"
        for prob in y_prob
    ]

    precision = precision_score(
        y_test, predictions, pos_label="Yes"
    )

    recall = recall_score(
        y_test, predictions, pos_label="Yes"
    )

    f1 = f1_score(
        y_test, predictions, pos_label="Yes"
    )

    print(
        f"Threshold: {threshold:.2f} | "
        f"Precision: {precision:.3f} | "
        f"Recall: {recall:.3f} | "
        f"F1: {f1:.3f}"
    )
```
